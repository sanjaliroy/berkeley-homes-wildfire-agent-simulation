# 04. Replicated evaluation with mean, SE and confidence intervals

Evaluates the runs produced by `03_run_simulation.ipynb`. Run once per model family, setting
`MODEL_FAMILY` below to match the simulation notebook.

Judging protocol, per family:
1. Full-simulation judge: every run and agent judged once (variants x replicates x agents).
2. Judge variance: replicate 1 of every condition re-judged `N_JUDGE_REPEATS` times at
   temperature 0, giving per-condition judge noise. Independent of the replicate count,
   so it runs even for a single-run matrix.
3. Per-intervention judge: one replicate of each variant (variants x agents x 6 events),
   feeding the appendix tables.

The replicate count is read from the runs on disk, not assumed. With one run per condition
the mean tables are still produced, but standard errors, delta confidence intervals and the
simulation-variance tables are skipped: a single run gives no run-to-run spread to estimate.
Set `N_REPLICATES > 1` in 03 to enable them.

Output, summary tables first and then full results:
- 6.2.1 Overall Results: per-dimension means with deltas against Baseline, mean and SE, 95% CIs on deltas
- 6.2.3 Model Comparison: Baseline against Budget
- 6.2.4 Cost and Latency
- Judge variance: mean and SE per condition across judge repeats
- CI build-up, worked step by step: replicate scores, then per-condition n/mean/SD/SE, then delta
  95% CI (per-comparison Welch plus family-wise Dunnett)
- Appendix: full-simulation scores by agent, Baseline scoring by criterion by agent, per-intervention
  by agent

The first tab of the Excel export, `Report Tables`, carries the report-formatted 6.2.1, 6.2.3 and
6.2.4 tables, the per-condition simulation-against-judge variance tables, and the CI build-up.

All judge scores are cached to CSV, so re-running a cell only judges what is missing.

## 2. Load simulation runs

```python
LABEL_RE = re.compile(rf'^{MODEL_FAMILY}_(?P<variant>.+)_rep(?P<rep>\d+)$')

def parse_jsonl(path: Path) -> list:
    entries = []
    with open(path, encoding='utf-8') as f:
        for line in f:
            line = line.strip()
            if line:
                entries.append(json.loads(line))
    return entries

# {run_label: {run_label, variant, rep, run_id, run_config, decisions, run_summary, source_file}}
runs = {}
for path in sorted(RUNS_DIR.glob('*.jsonl')):
    entries = parse_jsonl(path)
    run_config = next((e for e in entries if e.get('entry_type') == 'run_config'), None)
    if not run_config:
        continue
    label = run_config.get('run_label', '')
    m = LABEL_RE.match(label)
    if not m:
        continue   # not part of this family's experiment
    run_summary = next((e for e in entries if e.get('entry_type') == 'run_summary'), None)
    if run_summary is None:
        print(f'  SKIP {path.name} ({label}) — incomplete run (no run_summary)')
        continue
    if label in runs:
        prev = RUNS_DIR / runs[label]['source_file']
        if path.stat().st_mtime <= prev.stat().st_mtime:
            print(f'  SKIP {path.name} — older duplicate of {label} (keeping {prev.name})')
            continue
        print(f'  NOTE {label}: replacing {prev.name} with newer {path.name}')
    runs[label] = {
        'run_label':   label,
        'variant':     m['variant'],
        'rep':         int(m['rep']),
        'run_id':      path.stem,   # JSONL filename = run_id; keys the judge caches
        'run_config':  run_config,
        'decisions':   [e for e in entries if e.get('entry_type') == 'decision'],
        'run_summary': run_summary,
        'source_file': path.name,
    }

print(f'{len(runs)} completed runs loaded for family {MODEL_FAMILY!r}\n')
_status = pd.DataFrame([{'variant': r['variant'], 'rep': r['rep']} for r in runs.values()])
if not _status.empty:
    print(_status.assign(n=1).pivot_table(index='variant', columns='rep', values='n',
                                          aggfunc='count', fill_value=0).to_string())

def decisions_by_agent(run_data):
    by_agent = defaultdict(list)
    for e in sorted(run_data['decisions'], key=lambda e: e['tick']):
        by_agent[e['agent_id']].append(e)
    return by_agent

def judge_context(agent_id, entries):
    cfg = agent_configs.get(agent_id, {})
    seed_narrative = cfg.get('seed_narrative', entries[0].get('seed_personality', ''))
    memory_seeds   = cfg.get('memory_seeds', [])
    return seed_narrative, memory_seeds
```

## 3. Full-simulation judging, every run × agent, once

Every run × agent judged once (variants × replicates × agents). Cached by (run_id, agent): re-running only judges pairs that are missing.

```python
def load_cache(path: Path) -> pd.DataFrame:
    """Read a judge-score cache. An empty/headerless file (left behind by a run
    where every judge call failed) is treated as no cache rather than an error."""
    if not path.exists() or path.stat().st_size == 0:
        return pd.DataFrame()
    try:
        return pd.read_csv(path)
    except pd.errors.EmptyDataError:
        print(f'  {path.name} is empty — ignoring and re-judging from scratch')
        return pd.DataFrame()


def judge_fullsim_once(label, run_data, agent_id, entries):
    seed_narrative, memory_seeds = judge_context(agent_id, entries)
    all_decisions = [{'day': e['tick'], 'event_type': e['event_type'], 'intervention': e['intervention'],
                      'decision': e['decision'], 'reasoning': e['reasoning']} for e in entries]
    scores = judge_full_simulation(
        client_anthropic=client_anthropic, config=judge_config,
        seed_narrative=seed_narrative, memory_seeds=memory_seeds,
        all_decisions=all_decisions, client_openrouter=client_openrouter,
    )
    return {
        'run_label': label, 'run_id': run_data['run_id'],
        'variant': run_data['variant'], 'rep': run_data['rep'],
        'agent_id': agent_id, 'agent_display_name': entries[0]['agent_display_name'],
        'behavioral_plausibility':     scores.get('behavioral_plausibility'),
        'persona_consistency':         scores.get('persona_consistency'),
        'intervention_responsiveness': scores.get('intervention_responsiveness'),
        'bp_note': scores.get('note_plausibility'),
        'pc_note': scores.get('note_consistency'),
        'ir_note': scores.get('note_responsiveness'),
        'judge_model': JUDGE_MODEL,
    }


def run_judging(jobs, worker, cache_path, key_cols, max_workers=JUDGE_PARALLELISM):
    """Run judge calls in parallel with an on-disk cache. jobs: list of (key_tuple, callable)."""
    cache = load_cache(cache_path)
    if not cache.empty:
        parsed_ok = ~cache[DIMS].apply(pd.to_numeric, errors='coerce').isna().all(axis=1)
        if (~parsed_ok).any():
            print(f'  discarding {(~parsed_ok).sum()} cached rows with unparseable judge output — will re-judge')
            cache = cache[parsed_ok].reset_index(drop=True)
    have = set(map(tuple, cache[key_cols].itertuples(index=False))) if not cache.empty else set()
    todo = [(key, fn) for key, fn in jobs if key not in have]
    print(f'{len(jobs)} judge jobs | {len(jobs) - len(todo)} cached | {len(todo)} to run')
    if not todo:
        return cache
    new_rows, errors = [], []
    with ThreadPoolExecutor(max_workers=max_workers) as pool:
        futures = {pool.submit(fn): key for key, fn in todo}
        for i, future in enumerate(as_completed(futures), 1):
            key = futures[future]
            try:
                new_rows.append(future.result())
            except Exception as exc:
                errors.append((key, repr(exc)))
                print(f'  FAILED {key}: {exc!r}')
            if new_rows and (i % 10 == 0 or i == len(todo)):   # checkpoint the cache
                pd.concat([cache, pd.DataFrame(new_rows)], ignore_index=True).to_csv(cache_path, index=False)
            if i % 10 == 0 or i == len(todo):
                print(f'  {i}/{len(todo)} done')
    cache = pd.concat([cache, pd.DataFrame(new_rows)], ignore_index=True) if new_rows else cache
    if not cache.empty:                 # never write a headerless file — it poisons the cache
        cache.to_csv(cache_path, index=False)
    if errors:
        print(f'{len(errors)} failures — re-run this cell to retry them.')
    return cache


fullsim_jobs = []
for label, run_data in sorted(runs.items()):
    for agent_id, entries in decisions_by_agent(run_data).items():
        fullsim_jobs.append(((run_data['run_id'], agent_id),
                             (lambda l=label, rd=run_data, a=agent_id, e=entries:
                              judge_fullsim_once(l, rd, a, e))))

fullsim = run_judging(fullsim_jobs, None, FULLSIM_CACHE, ['run_id', 'agent_id'])
current_run_ids = {r['run_id'] for r in runs.values()}
fullsim = fullsim[fullsim['run_id'].isin(current_run_ids)].reset_index(drop=True)
fullsim[DIMS] = fullsim[DIMS].apply(pd.to_numeric, errors='coerce')
fullsim['overall'] = fullsim[DIMS].mean(axis=1)
_n_partial = int(fullsim[DIMS].isna().any(axis=1).sum())
if _n_partial:
    print(f'WARNING: {_n_partial} rows have missing dimension scores (partial judge parses)')
print(f'\nFull-simulation scores: {len(fullsim)} rows '
      f'({fullsim["variant"].nunique()} variants x {fullsim["rep"].nunique()} reps x '
      f'{fullsim["agent_id"].nunique()} agents)')
```

## 4. Judge variance: replicate 1 of each condition re-judged 5x

For each of the four conditions, re-judge one fixed trajectory (replicate 1) five times at
temperature 0 (variants x agents x N_JUDGE_REPEATS calls). Same trajectories and same judge, so any spread
here is judge noise from API nondeterminism rather than simulation noise. Reported as mean and SE
per condition.

```python
# Judge variance: hold ONE run per condition fixed (replicate 1) and re-judge it
# N_JUDGE_REPEATS times. SE across the repeats = judge (API) noise, per condition.
jv_runs = {}   # variant -> the fixed run we re-judge
for v in VARIANT_ORDER:
    reps = sorted((r['rep'], lbl) for lbl, r in runs.items() if r['variant'] == v)
    if reps:
        jv_runs[v] = runs[reps[0][1]]   # lowest available replicate index
assert jv_runs, 'No runs found — run the simulation notebook first.'
print(f'Judge-variance targets ({N_JUDGE_REPEATS}x each): '
      + ', '.join(f"{v} [{rd['run_id']}]" for v, rd in jv_runs.items()))


def judge_variance_once(run_data, repeat, agent_id, entries):
    row = judge_fullsim_once(run_data['run_label'], run_data, agent_id, entries)
    row['repeat'] = repeat
    return row


jv_jobs = []
for v, run_data in jv_runs.items():
    for repeat in range(1, N_JUDGE_REPEATS + 1):
        for agent_id, entries in decisions_by_agent(run_data).items():
            jv_jobs.append(((run_data['run_id'], repeat, agent_id),
                            (lambda rd=run_data, r=repeat, a=agent_id, e=entries:
                             judge_variance_once(rd, r, a, e))))

judge_var = run_judging(jv_jobs, None, JUDGEVAR_CACHE, ['run_id', 'repeat', 'agent_id'])
_jv_run_ids = {rd['run_id'] for rd in jv_runs.values()}
judge_var = judge_var[judge_var['run_id'].isin(_jv_run_ids)].reset_index(drop=True)
judge_var[DIMS] = judge_var[DIMS].apply(pd.to_numeric, errors='coerce')
judge_var['overall'] = judge_var[DIMS].mean(axis=1)
print(f'Judge-variance scores: {len(judge_var)} rows '
      f'({judge_var["variant"].nunique()} conditions x {N_JUDGE_REPEATS} repeats x '
      f'{judge_var["agent_id"].nunique()} agents)')
```

## 5. Per-intervention judging, one replicate per variant

Feeds the appendix tables (per-criterion and per-intervention breakdowns).

```python
def judge_pi_once(label, run_data, entry):
    seed_narrative, memory_seeds = judge_context(entry['agent_id'], [entry])
    scores = judge_intervention(
        client_anthropic=client_anthropic, config=judge_config,
        seed_narrative=seed_narrative, memory_seeds=memory_seeds,
        intervention=entry['intervention'], decision=entry['decision'],
        reasoning=entry['reasoning'], client_openrouter=client_openrouter,
    )
    return {
        'run_label': label, 'run_id': run_data['run_id'],
        'variant': run_data['variant'], 'rep': run_data['rep'],
        'agent_id': entry['agent_id'], 'agent_display_name': entry['agent_display_name'],
        'day': entry['tick'], 'event_type': entry['event_type'],
        'behavioral_plausibility':     scores.get('behavioral_plausibility'),
        'persona_consistency':         scores.get('persona_consistency'),
        'intervention_responsiveness': scores.get('intervention_responsiveness'),
        'bp_note': scores.get('note_plausibility'),
        'pc_note': scores.get('note_consistency'),
        'ir_note': scores.get('note_responsiveness'),
        'judge_model': JUDGE_MODEL,
    }


pi_jobs = []
for variant in VARIANT_ORDER:
    reps = sorted(r['rep'] for r in runs.values() if r['variant'] == variant)
    if not reps:
        print(f'  WARNING: no runs for {variant} — skipping')
        continue
    label = f'{MODEL_FAMILY}_{variant}_rep{reps[0]}'
    run_data = runs[label]
    for entry in sorted(run_data['decisions'], key=lambda e: (e['agent_id'], e['tick'])):
        pi_jobs.append(((run_data['run_id'], entry['agent_id'], entry['tick']),
                        (lambda l=label, rd=run_data, e=entry: judge_pi_once(l, rd, e))))

interventions = run_judging(pi_jobs, None, INTERVENTION_CACHE, ['run_id', 'agent_id', 'day'])
interventions = interventions[interventions['run_id'].isin({r['run_id'] for r in runs.values()})].reset_index(drop=True)
interventions[DIMS] = interventions[DIMS].apply(pd.to_numeric, errors='coerce')
interventions['overall'] = interventions[DIMS].mean(axis=1)
print(f'Per-intervention scores: {len(interventions)} rows')
```

## 6. Statistics helpers

- Replicate-level score = mean over the agents (each agent's score = the full-simulation judge output).
- Mean ± SE per variant = across the replicate-level values (needs more than one replicate; skipped otherwise).
- Delta CIs vs Baseline: Welch's t per comparison (unequal variances), plus a family-wise
  Dunnett CI across all three ablations-vs-Baseline together (multiple-comparison control).

```python
def mean_se(values):
    x = np.asarray(values, dtype=float)
    x = x[~np.isnan(x)]
    if len(x) == 0:
        return np.nan, np.nan
    se = x.std(ddof=1) / np.sqrt(len(x)) if len(x) > 1 else np.nan
    return x.mean(), se


def welch_ci(a, b, conf=0.95):
    """CI for mean(a) - mean(b), Welch's t. Returns (delta, lo, hi)."""
    a = np.asarray(a, dtype=float); b = np.asarray(b, dtype=float)
    a, b = a[~np.isnan(a)], b[~np.isnan(b)]
    d = a.mean() - b.mean()
    va, vb = a.var(ddof=1) / len(a), b.var(ddof=1) / len(b)
    se = np.sqrt(va + vb)
    if se == 0:
        return d, d, d
    df = (va + vb) ** 2 / (va ** 2 / (len(a) - 1) + vb ** 2 / (len(b) - 1))
    tcrit = st.t.ppf(0.5 + conf / 2, df)
    return d, d - tcrit * se, d + tcrit * se


# Replicate-level scores: one row per (variant, rep), columns = dims + overall
rep_level = (fullsim.groupby(['variant', 'rep'])[DIMS + ['overall']]
             .mean().reset_index())
rep_level['variant'] = pd.Categorical(rep_level['variant'], VARIANT_ORDER, ordered=True)
rep_level = rep_level.sort_values(['variant', 'rep']).reset_index(drop=True)

present_variants = [v for v in VARIANT_ORDER if v in set(rep_level['variant'])]

def _vals(variant, col):
    return rep_level.loc[rep_level['variant'] == variant, col].values

def _model_for(variant):
    for r in runs.values():
        if r['variant'] == variant:
            return r['run_config'].get('decision_model', '?')
    return '?'

# Friendly model names for the report tables (fall back to the raw id if unmapped)
MODEL_DISPLAY = {
    'claude-sonnet-4-6':          'Claude Sonnet 4.6',
    'claude-haiku-4-5-20251001':  'Claude Haiku 4.5',
    'claude-opus-4-6':            'Claude Opus 4.6',
    'openai/gpt-5.4':             'GPT-5.4',
    'openai/gpt-5.4-mini':        'GPT-5.4 mini',
    'openai/gpt-5.4-nano':        'GPT-5.4 nano',
}

def _model_display(variant):
    return MODEL_DISPLAY.get(_model_for(variant), _model_for(variant))


# Replicate count is derived from the runs on disk, never assumed. Labels, captions
# and which sections run all key off these values.
N_REPS = int(rep_level.groupby('variant', observed=True).size().max()) if len(rep_level) else 0
REPLICATED = N_REPS > 1
N_AGENTS = int(fullsim.groupby(['variant', 'rep'], observed=True).size().max()) if len(fullsim) else 0


def _reps_phrase(n=None):
    """'a single run' / '5 replicate runs', for captions."""
    n = N_REPS if n is None else n
    return 'a single run' if n == 1 else f'{n} replicate runs'


def _se_suffix():
    """' ± SE' only when there is more than one run to take an SE over."""
    return ' ± SE' if REPLICATED else ''


def _sd(vals):
    v = np.asarray(vals, dtype=float); v = v[~np.isnan(v)]
    return v.std(ddof=1) if len(v) > 1 else np.nan


def _fmt_msd(vals):
    """'mean ± SD' across repeated values; bare mean when there is only one."""
    v = np.asarray(vals, dtype=float); v = v[~np.isnan(v)]
    if len(v) == 0:
        return 'n/a'
    if len(v) == 1:
        return f'{v.mean():.2f}'
    return f'{v.mean():.2f} ± {v.std(ddof=1):.3f}'


def _fmt_mse(vals):
    """'mean ± SE' across repeated values (SE = SD/sqrt(n)); bare mean when n == 1."""
    v = np.asarray(vals, dtype=float); v = v[~np.isnan(v)]
    if len(v) == 0:
        return 'n/a'
    if len(v) == 1:
        return f'{v.mean():.2f}'
    return f'{v.mean():.2f} ± {v.std(ddof=1) / np.sqrt(len(v)):.3f}'


print(f'Replicates per condition: {N_REPS}   agents per run: {N_AGENTS}')
if not REPLICATED:
    print('Single run per condition: SE, delta CIs and simulation-variance tables are '
          'skipped (no run-to-run spread to estimate).')
print()
print(rep_level.round(3).to_string(index=False))
```

## 6.2.1 Overall Results

Mean score across agents, averaged over replicate runs. Each agent's complete 60-day trajectory
evaluated in one judge call. Deltas vs Baseline in parentheses.

With a single run per condition the mean table is still produced; the ± SE and delta-CI
sub-tables are skipped, because one run gives no run-to-run spread to estimate.

```python
base_means = {c: _vals('Baseline', c).mean() for c in DIMS + ['overall']}

rows = []
for v in present_variants:
    row = {'Condition': DISPLAY_NAMES[v]}
    for c in DIMS + ['overall']:
        m = _vals(v, c).mean()
        cell = f'{m:.2f}'
        if v != 'Baseline':
            cell += f' ({m - base_means[c]:+.2f})'
        row[DIM_LABELS.get(c, 'Mean')] = cell
    rows.append(row)
t621 = pd.DataFrame(rows).set_index('Condition')
print(f'6.2.1 OVERALL RESULTS (mean across {_reps_phrase()}; delta vs Baseline in parens)')
print('=' * 100)
print(t621.to_string())

# Mean ± SE and the delta CIs both need more than one run per condition.
if REPLICATED:
    rows = []
    for v in present_variants:
        row = {'Condition': DISPLAY_NAMES[v]}
        for c in DIMS + ['overall']:
            row[DIM_LABELS.get(c, 'Mean')] = _fmt_mse(_vals(v, c))
        rows.append(row)
    print(f'\n6.2.1b MEAN ± SE ACROSS {N_REPS} REPLICATES '
          f'(replicate = mean over {N_AGENTS} agents)')
    print('=' * 100)
    print(pd.DataFrame(rows).set_index('Condition').to_string())

    # Delta vs Baseline with 95% CI, the significance test for each ablation effect
    rows = []
    for v in present_variants:
        if v == 'Baseline':
            continue
        d, lo, hi = welch_ci(_vals(v, 'overall'), _vals('Baseline', 'overall'))
        rows.append({'Condition': DISPLAY_NAMES[v],
                     'Delta (overall, vs Baseline)': f'{d:+.3f}',
                     '95% CI': f'[{lo:+.3f}, {hi:+.3f}]',
                     'CI excludes 0': 'YES — significant' if (lo > 0 or hi < 0) else 'no'})
    print('\n6.2.1c DELTA vs BASELINE — 95% CI (Welch t across replicate-level means)')
    print('=' * 100)
    print(pd.DataFrame(rows).set_index('Condition').to_string())
    print('\nA CI that excludes 0 means the ablation effect is distinguishable from '
          'run-to-run noise at the 95% level.')
else:
    print('\n6.2.1b / 6.2.1c SKIPPED — single run per condition.')
    print('=' * 100)
    print('Standard errors and delta confidence intervals need at least two runs per')
    print('condition to estimate run-to-run spread.')
    print('Set N_REPLICATES > 1 in 03_run_simulation.ipynb to enable these sections.')
```

## Judge variance: replicate 1 of each condition judged 5x

For each condition, hold one trajectory fixed (replicate 1) and re-judge it five times at
temperature 0. Per repeat, take the mean over agents, then report mean and SE across the five
repeats, per condition. Comparing this judge SE against the simulation SE across the five replicate
runs shows whether the observed spread is genuine run-to-run behaviour or judge nondeterminism.

```python
# Judge variance per condition: average over agents per (condition, repeat),
# then mean +- SE across the N_JUDGE_REPEATS repeats.
jv_per_repeat = (judge_var.groupby(['variant', 'repeat'])[DIMS + ['overall']]
                 .mean().reset_index())
jv_variants = [v for v in VARIANT_ORDER if v in set(jv_per_repeat['variant'])]

def _jv_vals(variant, col):
    return jv_per_repeat.loc[jv_per_repeat['variant'] == variant, col].values

print(f'JUDGE VARIANCE - replicate 1 of each condition judged {N_JUDGE_REPEATS}x '
      f'by {JUDGE_MODEL} @ temp 0  (mean +- SE across repeats)')
print('=' * 95)
rows = []
for v in jv_variants:
    row = {'Condition': DISPLAY_NAMES[v]}
    for c in DIMS + ['overall']:
        m, se = mean_se(_jv_vals(v, c))
        row[DIM_LABELS.get(c, 'Overall')] = f'{m:.2f} +- {se:.3f}'
    rows.append(row)
print(pd.DataFrame(rows).set_index('Condition').to_string())

if REPLICATED:
    print(f'\nSE COMPARISON (overall): simulation ({N_REPS} runs) vs judge '
          f'({N_JUDGE_REPEATS} re-judges), per condition')
    print('=' * 95)
    rows = []
    for v in jv_variants:
        sim_se = mean_se(_vals(v, 'overall'))[1]
        jud_se = mean_se(_jv_vals(v, 'overall'))[1]
        rows.append({'Condition': DISPLAY_NAMES[v],
                     'Simulation SE': f'{sim_se:.3f}', 'Judge SE': f'{jud_se:.3f}'})
    print(pd.DataFrame(rows).set_index('Condition').to_string())
else:
    print('\nSE COMPARISON SKIPPED — single run per condition, so there is no simulation')
    print('SE to compare the judge SE against. The judge-variance table above still holds:')
    print('it re-scores one fixed trajectory, so it measures judge noise at any replicate count.')
```

```python
# Label + formatting helpers
def _cond_with_model(v):
    """'Baseline (Claude Sonnet 4.6)' / 'Ablation 1 (No Reflection - Claude Sonnet 4.6)'."""
    disp, model = DISPLAY_NAMES[v], _model_display(v)
    return f'{disp[:-1]} - {model})' if disp.endswith(')') else f'{disp} ({model})'

DIM_COLS = [DIM_LABELS[d] for d in DIMS] + ['Mean']   # header labels for the 4 score columns
_COLS = DIMS + ['overall']                             # matching dataframe columns

# 6.2.1 Overall Results (means, deltas vs Baseline)
rows = []
for v in present_variants:
    row = {}
    for c, label in zip(_COLS, DIM_COLS):
        m = _vals(v, c).mean()
        row[label] = f'{m:.2f}' if v == 'Baseline' else f'{m:.2f} ({m - base_means[c]:+.2f})'
    rows.append(row)
t_overall = pd.DataFrame(rows, index=[DISPLAY_NAMES[v] for v in present_variants])
t_overall.index.name = 'Condition'

# 6.2.3 Model Comparison (Baseline vs Budget, plain means, friendly names)
rows, idx = [], []
for v in ['Baseline', 'Budget']:
    if v not in present_variants:
        continue
    rows.append({label: f'{_vals(v, c).mean():.2f}' for c, label in zip(_COLS, DIM_COLS)})
    idx.append(_cond_with_model(v))
t_model = pd.DataFrame(rows, index=idx)
t_model.index.name = 'Condition'

# 6.2.4 Cost and Latency
rows, idx = [], []
for v in present_variants:
    summaries = [r['run_summary'] for r in runs.values() if r['variant'] == v]
    q    = _vals(v, 'overall').mean()
    cost = np.nanmean([s.get('agent_cost_usd', np.nan) for s in summaries])
    lat  = np.nanmean([s.get('latency_seconds', np.nan) for s in summaries])
    rows.append({'Quality (overall, 1–5)': f'{q:.2f}',
                 'Agent cost / run (USD)': f'${cost:.2f}',
                 'Latency / run (seconds)': f'{lat:.0f}'})
    idx.append(_cond_with_model(v))
t_cost = pd.DataFrame(rows, index=idx)
t_cost.index.name = 'Condition'

# ── Simulation variance: mean ± SE across the replicate runs, per condition ──
rows = []
for v in present_variants:
    rows.append({label: _fmt_mse(_vals(v, c)) for c, label in zip(_COLS, DIM_COLS)})
t_simvar = pd.DataFrame(rows, index=[DISPLAY_NAMES[v] for v in present_variants])
t_simvar.index.name = 'Condition'

# Judge variance: replicate 1 of each condition, mean ± SE across repeats
rows = []
for v in jv_variants:
    rows.append({label: _fmt_mse(_jv_vals(v, c)) for c, label in zip(_COLS, DIM_COLS)})
t_judgevar = pd.DataFrame(rows, index=[DISPLAY_NAMES[v] for v in jv_variants])
t_judgevar.index.name = 'Condition'

# Variance comparison: simulation SE vs judge SE per condition (overall)
rows = []
for v in jv_variants:
    sim_se = mean_se(_vals(v, 'overall'))[1]
    jud_se = mean_se(_jv_vals(v, 'overall'))[1]
    ratio = (jud_se / sim_se) if (sim_se and not np.isnan(sim_se) and sim_se != 0) else np.nan
    rows.append({f'Simulation SE ({N_REPS} runs)': f'{sim_se:.3f}',
                 f'Judge SE ({N_JUDGE_REPEATS} re-judges)': f'{jud_se:.3f}',
                 'Judge / Sim ratio': ('n/a' if np.isnan(ratio) else f'{ratio:.2f}')})
t_varcmp = pd.DataFrame(rows, index=[DISPLAY_NAMES[v] for v in jv_variants])
t_varcmp.index.name = 'Condition'

# CI build-up ①: replicate-level overall scores (raw inputs)
_ci_piv = rep_level.pivot(index='variant', columns='rep', values='overall').reindex(present_variants)
t_ci_values = pd.DataFrame(index=[DISPLAY_NAMES[v] for v in present_variants])
t_ci_values.index.name = 'Condition'
for r in sorted(_ci_piv.columns):
    t_ci_values[f'Rep {int(r)}'] = [f'{x:.3f}' if pd.notna(x) else 'n/a' for x in _ci_piv[r].values]

# CI build-up ②: per-condition summary (n, mean, SD, SE)
rows = []
for v in present_variants:
    vals = _vals(v, 'overall'); n = len(vals)
    sd = vals.std(ddof=1) if n > 1 else float('nan')
    se = sd / np.sqrt(n) if n > 1 else float('nan')
    rows.append({'n (runs)': n, 'Mean': f'{vals.mean():.3f}',
                 'SD (ddof=1)': f'{sd:.3f}', 'SE = SD/√n': f'{se:.3f}'})
t_ci_summary = pd.DataFrame(rows, index=[DISPLAY_NAMES[v] for v in present_variants])
t_ci_summary.index.name = 'Condition'

# CI build-up ③: per-comparison (Welch) + family-wise (Dunnett) delta CIs
def _welch_chain(vc, vb):
    nb_, nc_ = len(vb), len(vc)
    seb = vb.std(ddof=1) / np.sqrt(nb_)
    sec = vc.std(ddof=1) / np.sqrt(nc_)
    delta = vc.mean() - vb.mean()
    se_delta = np.sqrt(seb ** 2 + sec ** 2)
    if se_delta == 0 or nb_ < 2 or nc_ < 2:
        return delta, se_delta, float('nan'), float('nan'), 0.0, delta, delta
    df_ = (seb ** 2 + sec ** 2) ** 2 / (seb ** 4 / (nb_ - 1) + sec ** 4 / (nc_ - 1))
    tstar = st.t.ppf(0.975, df_)
    margin = tstar * se_delta
    return delta, se_delta, df_, tstar, margin, delta - margin, delta + margin

# Family-wise Dunnett: all treatments vs the single Baseline control, simultaneously.
_treat_variants = [v for v in present_variants if v != 'Baseline']
_dunnett = {}
if _treat_variants and all(len(_vals(v, 'overall')) > 1 for v in _treat_variants + ['Baseline']):
    try:
        from scipy.stats import dunnett as _dunnett_test
        _dres = _dunnett_test(*[_vals(v, 'overall') for v in _treat_variants],
                              control=_vals('Baseline', 'overall'))
        _dci = _dres.confidence_interval(confidence_level=0.95)
        _dunnett = {v: (float(_dci.low[i]), float(_dci.high[i]))
                    for i, v in enumerate(_treat_variants)}
    except Exception as _e:
        print(f'[warn] Dunnett unavailable ({_e!r}); family-wise columns show n/a.')

rows, idx = [], []
_vb = _vals('Baseline', 'overall')
for v in present_variants:
    if v == 'Baseline':
        continue
    delta, se_d, df_, tstar, margin, lo, hi = _welch_chain(_vals(v, 'overall'), _vb)
    dlo, dhi = _dunnett.get(v, (np.nan, np.nan))
    rows.append({
        'Δ (cond − Base)':       f'{delta:+.3f}',
        'SE(Δ)=√(SE²c+SE²b)':    f'{se_d:.3f}',
        'df (Welch)':            ('n/a' if np.isnan(df_) else f'{df_:.2f}'),
        't* (0.975)':            ('n/a' if np.isnan(tstar) else f'{tstar:.3f}'),
        'Margin=t*·SE(Δ)':       f'{margin:.3f}',
        '95% CI (Welch)':        f'[{lo:+.3f}, {hi:+.3f}]',
        'Excludes 0 (per-comp)': ('YES' if (lo > 0 or hi < 0) else 'no'),
        'Dunnett 95% CI (FW)':   ('n/a' if np.isnan(dlo) else f'[{dlo:+.3f}, {dhi:+.3f}]'),
        'Excludes 0 (FW)':       ('n/a' if np.isnan(dlo) else ('YES' if (dlo > 0 or dhi < 0) else 'no')),
    })
    idx.append(f'{DISPLAY_NAMES[v]} − Baseline')
t_ci_delta = pd.DataFrame(rows, index=idx)
t_ci_delta.index.name = 'Comparison'

# Assemble (title, dataframe, caption), order matches the Excel tab.
# Headline tables always appear. Variance and CI tables are omitted entirely when
# there is one run per condition — an empty table with 'n/a' in every cell invites
# the reader to treat a missing statistic as a computed one.
_agents_reps = (f'{N_AGENTS} agents' if not REPLICATED
                else f'{N_AGENTS} agents × {N_REPS} replicate runs '
                     f'({N_AGENTS * N_REPS} full-simulation judge calls per condition)')

REPORT_TABLES = [
    ('6.2.1 Overall Results', t_overall,
     f'Mean score across {_agents_reps}. Each agent\u2019s complete 60-day trajectory is '
     'evaluated in one judge call, capturing narrative coherence and cross-event '
     'integration. Deltas vs Baseline in parentheses.'),
    ('6.2.3 Model Comparison', t_model,
     'Full-simulation judge scores by dimension (BP, PC, IR) for Baseline vs Budget, means '
     f'across {_agents_reps}. Shows which dimensions are most affected by model choice.'),
    ('6.2.4 Cost and Latency', t_cost,
     'Quality = full-simulation holistic judge overall score (mean of BP, PC, IR across '
     'agents' + (' and replicates' if REPLICATED else '') + '). Agent cost and latency cover '
     'decision and reflection calls only'
     + (f', averaged across the {N_REPS} replicate runs.' if REPLICATED else '.')),
]

if REPLICATED:
    REPORT_TABLES += [
        ('Simulation Variance \u2014 mean \u00b1 SE (per condition)', t_simvar,
         f'Mean \u00b1 SE of the score across the {N_REPS} replicate runs of each condition '
         f'(SE = SD/\u221a{N_REPS}, agent-mean per run). Seed memories and the judge are held '
         'fixed, so this is run-to-run simulation stochasticity (decision/reflection '
         'sampling). This is the input variance for the delta CI.'),
        ('Judge Variance \u2014 mean \u00b1 SE (per condition)', t_judgevar,
         f'Replicate 1 of each condition re-judged {N_JUDGE_REPEATS}\u00d7 at temperature 0 '
         '\u2014 identical trajectories, same judge. Mean \u00b1 SE across the judge repeats '
         '(agent-mean per repeat). Isolates judge (API) nondeterminism, per condition.'),
        ('Variance Comparison \u2014 Simulation vs Judge (overall)', t_varcmp,
         'Standard error of the overall score from two sources, per condition: simulation '
         f'(SE across the {N_REPS} replicate runs) vs judge (SE across the '
         f'{N_JUDGE_REPEATS} re-judges of replicate 1). Judge/Sim ratio \u226a 1 means most '
         'spread is genuine run-to-run behaviour, not judge noise.'),
        ('CI build-up \u2460 \u2014 Replicate-level overall scores', t_ci_values,
         f'Overall score (mean of BP, PC, IR across the {N_AGENTS} agents) for each of the '
         f'{N_REPS} replicate runs per condition. These are the raw inputs to every statistic '
         'below. Displayed to 3 dp; the CI is computed from full-precision values.'),
        ('CI build-up \u2461 \u2014 Per-condition summary', t_ci_summary,
         f'Per condition, across the {N_REPS} replicate runs: n = number of runs, Mean, '
         'SD (sample, ddof=1), SE = SD/\u221an (standard error of the condition mean). '
         'These feed the delta CI below.'),
        ('CI build-up \u2462 \u2014 Delta 95% CI (per-comparison + family-wise)', t_ci_delta,
         'Each ablation\u2019s overall-score delta vs Baseline. Per-comparison (Welch): '
         '\u0394 = mean(cond) \u2212 mean(Baseline); SE(\u0394) = \u221a(SE\u00b2cond + '
         'SE\u00b2Baseline); df = Welch\u2013Satterthwaite; t* = 0.975 quantile; '
         'Margin = t*\u00b7SE(\u0394); CI = \u0394 \u00b1 Margin (each interval 95% on its '
         'own). Family-wise (Dunnett): a single simultaneous 95% CI across all treatments-vs-'
         'Baseline, so the 95% covers the whole set at once. Report the family-wise verdict '
         'when claiming significance across multiple ablations.'),
    ]
else:
    print(f'[note] {_reps_phrase().capitalize()} per condition \u2014 variance and CI tables '
          'omitted from the report sheet.\n')

print('\n' + '#' * 100)
print('REPORT-FORMATTED TABLES  (written to Excel tab 1: "Report Tables")')
print('#' * 100)
for title, df, caption in REPORT_TABLES:
    print(f'\n{title}')
    print('=' * 100)
    print(df.to_string())
    print(f'({caption})')
```

```python
timestamp = datetime.now().strftime('%Y%m%d_%H%M%S')
export_path = EVAL_DIR / f'evaluation_v2_{MODEL_FAMILY}_{timestamp}.xlsx'

cost_rows = []
for label, r in sorted(runs.items()):
    cost_rows.append({'run_label': label, 'variant': r['variant'], 'rep': r['rep'],
                      'agent_model': r['run_config'].get('decision_model'),
                      'agent_cost_usd': r['run_summary'].get('agent_cost_usd'),
                      'latency_seconds': r['run_summary'].get('latency_seconds')})

from openpyxl.styles import Font, PatternFill, Alignment, Border, Side
from openpyxl.utils import get_column_letter

_THIN = Side(style='thin', color='000000')
_BORDER = Border(left=_THIN, right=_THIN, top=_THIN, bottom=_THIN)
_HEADER_FILL = PatternFill('solid', fgColor='D9D9D9')

def write_report_table(ws, start_row, title, df, caption, index_header):
    """Write one report-styled table (bold title, grey header, borders, italic caption)."""
    r = start_row
    ws.cell(r, 1, title).font = Font(bold=True, size=12)
    r += 1
    cols = [index_header] + list(df.columns)
    for j, name in enumerate(cols, start=1):
        c = ws.cell(r, j, name)
        c.font = Font(bold=True)
        c.fill = _HEADER_FILL
        c.border = _BORDER
        c.alignment = Alignment(horizontal='center', vertical='center', wrap_text=True)
    r += 1
    for idx, row_vals in zip(df.index, df.itertuples(index=False)):
        c0 = ws.cell(r, 1, str(idx))
        c0.border = _BORDER
        c0.alignment = Alignment(horizontal='left', vertical='center')
        for j, val in enumerate(row_vals, start=2):
            c = ws.cell(r, j, val)
            c.border = _BORDER
            c.alignment = Alignment(horizontal='center', vertical='center')
        r += 1
    cap = ws.cell(r, 1, caption)
    cap.font = Font(italic=True, size=9)
    cap.alignment = Alignment(wrap_text=True, vertical='top')
    r += 1
    # Column widths (widest requirement across all tables wins)
    ws.column_dimensions['A'].width = max(ws.column_dimensions['A'].width or 0, 48)
    for j in range(2, len(cols) + 1):
        L = get_column_letter(j)
        ws.column_dimensions[L].width = max(ws.column_dimensions[L].width or 0, 24)
    return r + 2   # blank gap before the next table


_EXPLAINER = [
    ('h1', 'How to read this sheet'),
    ('p',  'What this is: a fidelity check of an LLM agent simulation (generative agents, '
           'Park et al. 2023). Each condition \u2014 Baseline plus three ablations \u2014 is run '
           + ('once' if not REPLICATED else f'{N_REPS} times (replicates)') + '. An LLM judge '
           'scores each agent\u2019s full 60-day trajectory on three 1\u20135 dimensions: '
           'Behavioral Plausibility (BP), Persona Consistency (PC), Intervention '
           'Responsiveness (IR). \u201cOverall\u201d is the mean of the three.'),
    ('p',  'Tables 6.2.1 / 6.2.3 / 6.2.4 are the headline results (quality, model comparison, '
           'cost & latency).'
           + (' The variance tables show how stable those numbers are. CI build-up '
              '\u2460\u2013\u2462 shows the significance test step by step so it can be '
              're-derived by hand.' if REPLICATED else '')),
]

if REPLICATED:
    _EXPLAINER += [
        ('h2', 'Why these methods'),
        ('p',  f'\u2022 Mean \u00b1 SE (SE = SD/\u221a{N_REPS}): SE is how precisely each '
               f'condition\u2019s average is pinned down across its {N_REPS} runs. We report SE, '
               'not SD, because the claim is about the average score, not one run.'),
        ('p',  '\u2022 95% CI on each ablation \u2212 Baseline gap, using Welch\u2019s t-test: '
               'run-to-run spread comes from the LLM\u2019s own sampling and can differ between '
               'conditions (e.g. scores bunching near the 5.0 ceiling). Welch allows unequal '
               'spread and is the recommended default; the pooled t-test would wrongly assume '
               'equal spread. \u201cExcludes 0\u201d means the gap is bigger than run-to-run '
               'noise at 95% confidence.'),
        ('p',  '\u2022 Multiple comparisons: we test three ablations against Baseline, so next '
               'to each per-comparison 95% CI we also report a family-wise (Dunnett) 95% CI. '
               'Three separate 95% tests carry ~14% chance of a false \u201csignificant\u201d '
               'somewhere, not 5%. Dunnett widens the intervals just enough to keep 95% '
               'confidence across all three comparisons at once, and is the standard test for '
               'several treatments vs one control. Use the \u201cfamily-wise\u201d verdict for '
               'significance claims.'),
        ('p',  f'\u2022 n = {N_REPS} runs, not {N_REPS * N_AGENTS} agent-scores: the {N_AGENTS} '
               'agents within a run are not independent, so each run is collapsed to its mean '
               f'first. Counting {N_REPS * N_AGENTS} scores as independent would be a fake '
               'sample size and overstate significance.'),
        ('p',  f'\u2022 t-distribution, not the normal (z): with only {N_REPS} runs the interval '
               'is widened to stay honest about small-sample uncertainty.'),
        ('p',  f'\u2022 The CI is on the simulation only. The judge is re-run '
               f'{N_JUDGE_REPEATS}\u00d7 per condition just to confirm it adds little noise (a '
               'measurement check) \u2014 not to test a hypothesis \u2014 so there is no judge '
               'CI. Judge noise from the single scoring is already included in each run\u2019s '
               'score.'),
    ]
else:
    _EXPLAINER += [
        ('h2', 'Why there are no confidence intervals'),
        ('p',  'Each condition was run once, so there is no run-to-run spread to estimate. '
               'Standard errors, delta confidence intervals and the significance tests are '
               'therefore omitted rather than shown as blanks.'),
        ('p',  'The judge-variance table is still meaningful: it re-scores one fixed '
               'trajectory several times, so it measures judge nondeterminism independently '
               'of how many times the simulation was run.'),
        ('p',  'To enable the full statistical treatment, set N_REPLICATES > 1 in '
               '03_run_simulation.ipynb and re-run both notebooks. Five replicates per '
               'condition supports Welch and Dunnett intervals.'),
    ]

def write_explainer(ws, start_row, blocks, ncols=8):
    """Write a wrapped, merged explainer block at the top of the sheet."""
    r = start_row
    for kind, text in blocks:
        ws.merge_cells(start_row=r, start_column=1, end_row=r, end_column=ncols)
        cell = ws.cell(r, 1, text)
        if kind == 'h1':
            cell.font = Font(bold=True, size=12)
        elif kind == 'h2':
            cell.font = Font(bold=True, size=10)
        else:
            cell.font = Font(size=10)
        cell.alignment = Alignment(wrap_text=True, vertical='top', horizontal='left')
        n_lines = max(1, -(-len(text) // 200))   # ceil(len / ~chars-per-line)
        ws.row_dimensions[r].height = 15 * n_lines + 4
        r += 1
    return r + 1   # blank gap before the first table


with pd.ExcelWriter(export_path, engine='openpyxl') as writer:
    wb = writer.book
    default_ws = wb.active   # openpyxl's blank default sheet — removed at the end
    ws = wb.create_sheet('Report Tables', 0)   # first tab
    row = 1
    ws.cell(row, 1, f'Evaluation report tables — {MODEL_FAMILY} family  '
                    f'(judge: {JUDGE_MODEL})').font = Font(bold=True, size=14)
    row += 2
    row = write_explainer(ws, row, _EXPLAINER)
    for title, df, caption in REPORT_TABLES:
        row = write_report_table(ws, row, title, df, caption, df.index.name)

    # Raw / detailed sheets after the report tab
    fullsim.to_excel(writer, sheet_name='FullSim Scores', index=False)
    rep_level.to_excel(writer, sheet_name='Replicate Level', index=False)
    judge_var.to_excel(writer, sheet_name='Judge Variance', index=False)
    interventions.to_excel(writer, sheet_name='Per-Intervention', index=False)
    pd.DataFrame(cost_rows).to_excel(writer, sheet_name='Cost & Latency', index=False)
    appx_a.to_excel(writer, sheet_name='Appx A FullSim by Agent')
    appx_b.to_excel(writer, sheet_name='Appx B Baseline Criteria')

    if default_ws is not None and default_ws.title == 'Sheet' and default_ws is not ws:
        wb.remove(default_ws)

print(f'Exported: {export_path}')
print(f'First tab "Report Tables" holds the 6.2.1 / 6.2.3 / 6.2.4 + variance tables.')

judge_usage = usage_tracker.to_dict(agent_model=JUDGE_MODEL, judge_model=JUDGE_MODEL)
print(f"\nJudging cost this session ({JUDGE_MODEL}): ${judge_usage['judge_cost_usd']:.4f} "
      f"({judge_usage['judge_tokens_in']:,} in / {judge_usage['judge_tokens_out']:,} out)")
```
